```python
import json
import pandas as pd
```

```python
with open("../sources/benchmark.json", "r") as benchmark_file:
    benchmark = json.load(benchmark_file)
```

```python
type_counts = {}
for item in benchmark:
    type_counts[item["type"]] = type_counts.get(item["type"], 0) + 1

for k, v in type_counts.items():

    print(f'{k} : {v}')
```

```python
task_counts = {}
for item in benchmark:
    task_counts[item["task"]] = task_counts.get(item["task"], 0) + 1

for k, v in task_counts.items():

    print(f'{k} : {v}')
```

```python
domain_counts = {}
for item in benchmark:
    domain_counts[item["news_domain"]] = domain_counts.get(item["news_domain"], 0) + 1

for k, v in domain_counts.items():

    print(f'{k} : {v}')
```

```python
type_task_counts = {}
for item in benchmark:
    type_task_counts[(item["type"], item["task"])] = type_task_counts.get((item["type"], item["task"]), 0) + 1

type_task_df = pd.Series(type_task_counts).unstack(level=0, fill_value=0)
type_task_df
```
